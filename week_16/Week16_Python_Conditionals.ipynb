## Recap: Logic gates → Python booleans- Gates output 0/1; Python uses `False`/`True`.- AND gate ≈ `and`; OR gate ≈ `or`; NOT gate ≈ `not`.- XOR (exactly one) can be built with `!=` or the `^` operator on booleans.

````python
# Truth tables as Python booleans (mirrors your logic gate tables)inputs = [(False, False), (False, True), (True, False), (True, True)]print("A	B	AND	OR	NOT A	XOR")for a, b in inputs:    and_out = a and b    or_out = a or b    not_out = not a    xor_out = (a != b)  # True when values differ    print(f"{a}	{b}	{and_out}	{or_out}	{not_out}	{xor_out}")
````

## `elif` for multiple casesUse `elif` when cases are mutually exclusive:```if condition_a:    ...elif condition_b:    ...else:    ...```The first true condition wins; later ones are skipped.

````python
# Example: categorize a temperature readingtemp_f = 68if temp_f < 50:    category = "cold"elif temp_f < 75:    category = "comfortable"elif temp_f < 90:    category = "warm"else:    category = "hot"print("Temperature:", temp_f, "->", category)
````
