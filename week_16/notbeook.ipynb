## Recap: Logic gates → Python booleans
- Gates output 0/1; Python uses `False`/`True`.
- AND gate ≈ `and`; OR gate ≈ `or`; NOT gate ≈ `not`.
- XOR (exactly one) can be built with `!=` or the `^` operator on booleans.


````python

# Truth tables as Python booleans (mirrors your logic gate tables)
inputs = [(False, False), (False, True), (True, False), (True, True)]

print("A\tB\tAND\tOR\tNOT A\tXOR")
for a, b in inputs:
    and_out = a and b
    or_out = a or b
    not_out = not a
    xor_out = (a != b)  # True when values differ
    print(f"{a}\t{b}\t{and_out}\t{or_out}\t{not_out}\t{xor_out}")
````

## Basic `if` / `else` structure
```
if condition_is_true:
    # do this block
else:
    # otherwise do this block
```
Indentation (4 spaces) tells Python which lines belong to the block.


````python

# Example: two-factor style check using and/or/not
password_ok = True
code_ok = False
account_locked = False

if account_locked:
    print("Account is locked. Try later or reset.")
elif password_ok and code_ok:
    print("Access granted.")
else:
    print("Access denied. Check password/code.")
````

## Truthy and falsy values
Besides `True`/`False`, Python treats some values as falsey: `0`, `0.0`, empty strings `""`, empty lists `[]`, empty dicts `{}`, and `None`. Everything else is truthy.

## Exercise 2: Build a toggle (XOR-like)
Create a function that returns the **new** light state when a button is pressed:
- If button is pressed, flip the current state.
- If not pressed, keep the current state.
Use booleans only—no numbers.


````python

# TODO: Complete the toggle function. Use clear variable names and comments.
def next_light_state(current_on: bool, button_pressed: bool) -> bool:
    "Return the new light state after a button press."
    if button_pressed:
        return not current_on  # flip state when pressed
    return current_on          # otherwise stay the same

# Quick checks (feel free to add more)
print(next_light_state(False, True))   # Expect True (off -> on)
print(next_light_state(True, True))    # Expect False (on -> off)
print(next_light_state(True, False))   # Expect True (no change)
````

## `elif` for multiple cases
Use `elif` when cases are mutually exclusive:
```
if condition_a:
    ...
elif condition_b:
    ...
else:
    ...
```
The first true condition wins; later ones are skipped.


````python

# Example: categorize a temperature reading
temp_f = 68

if temp_f < 50:
    category = "cold"
elif temp_f < 75:
    category = "comfortable"
elif temp_f < 90:
    category = "warm"
else:
    category = "hot"

print("Temperature:", temp_f, "->", category)
````
