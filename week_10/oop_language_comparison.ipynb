## Python Implementation
The class below models the CRC card we sketched for a `StudyGroup`. Run the cell, then explore suggested experiments.


```
from dataclasses import dataclass, field
from datetime import datetime, timedelta
from typing import List, Optional

@dataclass
class StudyGroup:
    """Track members, meetings, and shared notes for a study group."""

    topic: str
    members: List[str] = field(default_factory=list)
    facilitator: Optional[str] = None
    meeting_notes: List[str] = field(default_factory=list)
    last_meeting_at: Optional[datetime] = None

    def add_member(self, name: str) -> int:
        """Register a new member if they are not already in the roster."""
        if name not in self.members:
            self.members.append(name)
        return len(self.members)

    def schedule_meeting(self, when: datetime, agenda: str) -> None:
        """Record the next meeting time and capture the agenda as a note."""
        self.last_meeting_at = when
        self.meeting_notes.append(f"Agenda for {when:%b %d}: {agenda}")

    def summary(self) -> str:
        """Provide a single-line snapshot for announcements or logs."""
        headcount = len(self.members)
        facilitator = self.facilitator or "TBD"
        return f"StudyGroup(topic={self.topic}, members={headcount}, facilitator={facilitator})"


group = StudyGroup(topic="Discrete Math", facilitator="Amira")
print(group.summary())
print(f"Members after adding: {group.add_member('Luis')}")
print("Meeting history before scheduling:", group.meeting_notes)

next_week = datetime.now() + timedelta(days=7)
group.schedule_meeting(next_week, "Proof strategies showcase")
print(group.summary())
print("Latest note:", group.meeting_notes[-1])
```

## JavaScript Translation
Execute the next cell to run an equivalent class in Node.js directly from this notebook. Compare the output with the Python version.


```
%%bash
node <<'NODE'
class StudyGroup {
  constructor(topic, members = [], facilitator = null) {
    this.topic = topic;
    this.members = [...members];
    this.facilitator = facilitator;
    this.meetingNotes = [];
    this.lastMeetingAt = null;
  }

  addMember(name) {
    if (!this.members.includes(name)) {
      this.members.push(name);
    }
    return this.members.length;
  }

  scheduleMeeting(whenIso, agenda) {
    this.lastMeetingAt = whenIso;
    this.meetingNotes.push(`Agenda for ${whenIso}: ${agenda}`);
  }

  summary() {
    const facilitatorName = this.facilitator ?? 'TBD';
    return `StudyGroup(topic=${this.topic}, members=${this.members.length}, facilitator=${facilitatorName})`;
  }
}

const group = new StudyGroup('Discrete Math', [], 'Amira');
console.log(group.summary());
console.log(`Members after adding: ${group.addMember('Luis')}`);
console.log('Meeting history before scheduling:', group.meetingNotes);

const nextWeek = new Date(Date.now() + 7 * 24 * 60 * 60 * 1000).toISOString();
group.scheduleMeeting(nextWeek, 'Proof strategies showcase');
console.log(group.summary());
console.log('Latest note:', group.meetingNotes.at(-1));

NODE
```

### Syntax Checkpoint
| Concept | Python | JavaScript |
| --- | --- | --- |
| Define class | `class Name:` | `class Name {` |
| Constructor | `def __init__(self, ...)` | `constructor(...) { ... }` |
| Instance reference | `self.attribute` | `this.attribute` |
| Method separation | Indentation + `def` | Inside class body without `function` keyword |
| Default list argument | `members=None` then `members or []` | `members = []` in parameters + spread `[...]` |
| Private-ish fields | `_hidden` convention or `@property` | `#privateField` syntax or closures |


## Habit Tracker Challenge
1. Brainstorm responsibilities for a `HabitTracker` class (log entry, streak count, celebrate milestones).
2. Draft a quick CRC card: collaborators might include `ReminderService` or `DataStore`.
3. Implement the class in Python using the cell below; then port the logic to JavaScript by editing the node cell.


```
# TODO: Implement your HabitTracker class here
from __future__ import annotations
from dataclasses import dataclass, field
from datetime import date
from typing import Dict

@dataclass
class HabitTracker:
    name: str
    log: Dict[date, bool] = field(default_factory=dict)

    def log_entry(self, day: date, completed: bool = True) -> None:
        """Record whether the habit was completed on the given day."""
        self.log[day] = completed

    def current_streak(self, today: date | None = None) -> int:
        """Return the number of consecutive completed days ending with today."""
        today = today or date.today()
        streak = 0
        check_day = today
        while self.log.get(check_day):
            streak += 1
            check_day = check_day.fromordinal(check_day.toordinal() - 1)
        return streak


tracker = HabitTracker(name="Daily Reading")
for offset in range(3):
    tracker.log_entry(date.today().fromordinal(date.today().toordinal() - offset))
print("Current streak:", tracker.current_streak())
```
